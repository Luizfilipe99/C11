#Exercicio 4

```python
import numpy as np

np.random.seed(10)

mtz = np.random.randint(1, 51, [4, 4])

print(mtz.mean(axis=0))
print(mtz.mean(axis=1))

print(mtz.mean(axis=0).max())
print(mtz.mean(axis=1).max())


valores, contagens = np.unique(mtz, return_counts=True)

print("Quantidade de aparições de cada número:")
for num, qtd in zip(valores, contagens):
    print(f"O número {num} aparece {qtd} vez(es)")

numeros_duas_vezes = valores[contagens == 2]

print("\nNúmeros que aparecem 2 vezes:")
print(numeros_duas_vezes)
```
