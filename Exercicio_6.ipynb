```python
import pandas as pd
import numpy as np
```

#Exercicio 1

```python
labels0 = ['Java', 'C', 'Python']
labels1 = ['C', 'Python', 'Java']
dados0 = [16.25, 16.04, 9.85]
dados1 = [16.21, 12.12, 11.68]

seriesAno1 = pd.Series(index=labels0, data=dados0)
seriesAno2 = pd.Series(index=labels1, data=dados1)

print(seriesAno1)
print(seriesAno2)
```

#Execicio 2

```python
print(sum(seriesAno1))
print(sum(seriesAno2))
```

#Exercicio 3

```python
delta = seriesAno1 - seriesAno2
print(delta)
```

#Exercicio 4

```python
for i in range(0, len(delta)):
  if delta.iloc[i] > 0:
    print(delta.iloc[i])
```

#Exercicio 5

```python
delta2 = seriesAno2 + (delta * 2)

print(delta2)

print(delta2.nlargest(1))
```

#Exercicio 6

```python
from numpy._core.fromnumeric import mean
df = pd.DataFrame(
 index = ['A', 'B', 'C', 'D', 'E'],
 columns = ['W', 'X', 'Y', 'Z'],
 data = np.random.randint(1, 50, [5, 4])
)

lista0 = []

for i in range(0, 5):
  for t in range(0, 4):
    if df.iloc[i, t] > 30:
      lista0.append(df.iloc[i, t])

print(mean(lista0))
```
