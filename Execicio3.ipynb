```python
list = []

n = input('quantas pessoas serão? ')

for i in range(int(n)):
  i = {}
  i['nome'] = str(input('Digite o nome: '))
  i['idade'] = int(input('Digite a idade: '))
  i['sexo'] = str(input('Digite o sexo: '))
  list.append(i)

media = 0
for i in range(int(n)):
  media += list[i]['idade']

media = int(media / int(n))
print('Media de idade: ', media)

d = 0
for i in range(int(n)):
  if list[i]['sexo'] == 'f':
    d += 1

print('Numero de mulheres: ', d)

```
