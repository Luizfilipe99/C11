#Exercicio 8

```python
numero3 = int(input('entre com um numero: '))
numero4 = int(input('entre com outro numero: '))

print('soma: ')
if (numero3 + numero4) % 2 == 0:
  print('par')
else:
  print('impar')

print('subtração: ')
if (numero3 - numero4) % 2 == 0:
  print('par')
else:
  print('impar')

print('multiplicação: ')
if (numero3 * numero4) % 2 == 0:
  print('par')
else:
  print('impar')

print('divisão: ')
if (numero3 / numero4) % 2 == 0:
  print('par')
else:
  print('impar')

print('resto: ')
if (math.fmod(numero3, numero4)) % 2 == 0:
  print('par')
else:
  print('impar')

print('elevado: ')
if math.pow(numero3, numero4) % 2 == 0:
  print('par')
else:
  print('impar')
```
